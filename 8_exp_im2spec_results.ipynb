```python
import pickle
import numpy as np
import matplotlib.pyplot as plt
with open(f"outputs/2e_d_result.pkl", "rb") as f:
    d_result_loaded = pickle.load(f)

with open(f"outputs/2e_d_ratios.pkl", "rb") as f:
    d_ratios_loaded = pickle.load(f)   
```

```python
list_active, list_random, list_active2, list_active3 = [], [], [], []
for k, v in d_result_loaded.items():
    list_active.append(v[0])
    list_random.append(v[1])
    list_active2.append(v[2])
    list_active3.append(v[3])    
a, b, c, d = np.array(list_active), np.array(list_random), np.array(list_active2), np.array(list_active3)

mean_a, std_a = a.mean(axis=0), a.std(axis=0)
mean_b, std_b = b.mean(axis=0), b.std(axis=0)
mean_c, std_c = c.mean(axis=0), c.std(axis=0)
mean_d, std_d = d.mean(axis=0), d.std(axis=0)

steps = np.arange(0, a.shape[1])
labels = ["Active", "Random", "ActiveQC", "ActiveMT"]
colors = ["tab:blue", "tab:orange", "tab:green", "tab:purple"]

plt.figure(figsize=(8, 4))
plt.errorbar(steps, mean_a, yerr=std_a, fmt='-o', color=colors[0], label=labels[0], capsize=3, markersize=4)
plt.errorbar(steps, mean_b, yerr=std_b, fmt='-o', color=colors[1], label=labels[1], capsize=3, markersize=4)
plt.errorbar(steps, mean_c, yerr=std_c, fmt='-o', color=colors[2], label=labels[2], capsize=3, markersize=4)
plt.errorbar(steps, mean_d, yerr=std_d, fmt='-o', color=colors[3], label=labels[3], capsize=3, markersize=4)

plt.xlabel("Exploration Step")
plt.ylabel("Error (mean ± std)")
plt.title("Performance Comparison")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
steps = np.arange(0, a.shape[1])
labels = ["Active", "Random", "ActiveQC", "ActiveMT"]
colors = ["tab:blue", "tab:orange", "tab:green", "tab:purple"]

num_trials = a.shape[0]

for trial in range(num_trials):
    plt.figure(figsize=(6, 4))
    plt.plot(steps, a[trial], color=colors[0], label=labels[0], lw=2)
    plt.plot(steps, b[trial], color=colors[1], label=labels[1], lw=2)
    plt.plot(steps, c[trial], color=colors[2], label=labels[2], lw=2)
    plt.plot(steps, d[trial], color=colors[3], label=labels[3], lw=2)
    
    plt.xlabel("Exploration Step")
    plt.ylabel("Ratio")
    plt.title(f"Trial {trial+1}: Comparison of All Models")
    plt.legend()
    plt.grid(alpha=0.3)
    plt.tight_layout()
    plt.show()
```
